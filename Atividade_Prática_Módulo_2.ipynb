```python
num1_x = int(input("Digite o primeiro número:"))
num2_x = int(input("Digite o segundo número:"))
num3_x = int(input("Digite o terceiro número: "))
soma = num1_x + num2_x + num3_x
subtracao = num1_x - num2_x - num3_x
multiplicacao = num1_x * num2_x * num3_x
divisao = num1_x / num2_x / num3_x

print("A soma é:", soma)
print("A subtração é:", subtracao)
print("A multiplicação é:", multiplicacao)
print("A divisão é:", divisao)
```

```python
num1_x = int(input("Digite o primeiro número:"))
num2_x = int(input("Digite o segundo número:"))
num3_x = int(input("Digite o terceiro número: "))
soma = num1_x + num2_x + num3_x
media = num1_x  + num2_x + num3_x / 3
produto = num1_x * num2_x * num3_x

maior = num1_x
if num2_x > maior:
    maior = num2_x
if num3_x > maior:
    maior = num3_x
print("O maior número é:", maior)
menor = num1_x
if num2_x < menor:
    menor = num2_x
if num3_x < menor:
    menor = num3_x

diferenca = maior - menor

print("A soma é:", soma)
print("A média é:", media)
print("O produto é:", produto)
print("O maior número:", maior)
print("O menor número:", menor)
print("A diferença entre o maior e menor número é:", diferenca)
```
