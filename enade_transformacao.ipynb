```python
import urllib.request
import zipfile
import os
import ssl

url = 'https://download.inep.gov.br/microdados/microdados_enade_2023.zip'
zip_path = './enade2023/microdados_enade_2023.zip'

# Criando um contexto SSL que ignora a verificação de certificado
gcontext = ssl._create_unverified_context()

# Criando uma requisição com User-Agent para simular um navegador comum e evitar bloqueio
req = urllib.request.Request(
    url,
    headers={'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'}
)

try:
    print("Tentando baixar os dados do ENADE 2023 ignorando validação SSL...")
    with urllib.request.urlopen(req, context=gcontext) as response, open(zip_path, 'wb') as out_file:
        out_file.write(response.read())

    print("Download concluído! Extraindo os arquivos...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall('./enade2023')
    print("Concluído! Agora você pode rodar a célula de leitura de dados.")
except Exception as e:
    print(f"Ocorreu um erro no download: {e}")
    print("Se o bloqueio persistir, será necessário fazer o download manual no seu navegador e subir o zip para o painel lateral do Colab.")
```
