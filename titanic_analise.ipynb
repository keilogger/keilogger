```python
import kagglehub
import pandas as pd
import os

# Download do dataset
path = kagglehub.dataset_download("heptapod/titanic")

print("Path to dataset files:", path)

# Verificar arquivos disponíveis
print(os.listdir(path))
```

```python
arquivo = os.path.join(path, "train_and_test2.csv")

df = pd.read_csv(arquivo)

df.head()
```

```python
df.info()

df.describe()
```

```python
#Análise Exploratória e Limpeza dos Dados
```

```python
df.isnull().sum()
```

```python
df['Age'].fillna(df['Age'].median(), inplace=True)

df['Fare'].fillna(df['Fare'].median(), inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['2urvived'].value_counts()
```

```python
#Visualização dos Dados
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='2urvived')

plt.title('Sobreviventes x Não Sobreviventes')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='Sex', hue='2urvived')

plt.title('Sobrevivência por Sexo')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='Pclass', hue='2urvived')

plt.title('Sobrevivência por Classe')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=20)

plt.title('Distribuição de Idades')
plt.show()
```

```python
plt.figure(figsize=(18,10))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm')

plt.title('Mapa de Correlação')
plt.show()
```
